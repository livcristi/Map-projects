## Data Gathering

Code for downloading Wikipedia articles for different countries / states

```python
# Constants for operations
LANGUAGE = "en"
WIKI_ENDPOINT = f"https://{LANGUAGE}.wikipedia.org/w/api.php"
CONTACT_EMAIL = userdata.get('CONTACT_EMAIL') # This is an env var for Google Colab, you can replace it with an actual email address
USER_AGENT = f"eu-us-semantic-map/1.0 ({CONTACT_EMAIL})"
ARTICLES_FOLDER = "Articles"

# Some wiki articles are named differently than the country name from the geopandas dataset
TERRITORY_ARTICLE_MAPPING: dict[str, str] = {
    # European countries
    "Czechia": "Czech_Republic",
    "Ireland": "Republic_of_Ireland",
    "Vatican": "Vatican_City",
    "Bosnia_and_Herz.": "Bosnia_and_Herzegovina",
    # US States
    "Washington": "Washington_(state)",
    "New_York": "New_York_(state)",
    "Georgia": "Georgia_(U.S._state)",
    "District_of_Columbia": "Washington,_D.C.",
}

# Model names
MODEL_NAME_PRIMARY = "google/embeddinggemma-300m" # Is one of the best models available
MODEL_NAME_FALLBACK = "sentence-transformers/all-mpnet-base-v2" # Is a good trade-off for speed
MAX_TOKENS = 2048

# Logs are preffered over print statements
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger("eu_us_semantic_map")
```

```python
def create_directories() -> None:
    os.makedirs(ARTICLES_FOLDER, exist_ok=True)
```

```python
def retrieve_wikipedia_article(article_title: str) -> str:
    """
    Fetches a plain-text Wikipedia article based on its title.

    For details on how it works, visit the [Wikipedia API docs](https://www.mediawiki.org/wiki/Extension:TextExtracts#API).
    """

    headers: dict[str, str] = {
        "User-Agent": USER_AGENT,
        "Accept": "text/html,application/xhtml+xml,application/json",
        "Accept-Language": "en",
    }

    params: dict[str, str] = {
        "action": "query",
        "prop": "extracts",
        "format": "json",
        "explaintext": True,
        "titles": article_title,
        "formatversion": 2
    }

    try:
        response: Response = requests.get(WIKI_ENDPOINT, headers=headers, params=params, timeout=5)

        # Raise possible exceptions
        response.raise_for_status()

        raw_data: dict[str, Any] = response.json()
        pages_data: list[Any] = glom(raw_data, "query.pages", default=[])
        if not pages_data:
            raise ValueError(f"No pages found for title: {article_title}")
        return glom(pages_data[0], "extract", default="")
    except Exception as e:
        print(f"Error while retrieving Wikipedia article: {e}")
        raise
```

```python
def store_article_data(article_title: str, article_folder: str = ARTICLES_FOLDER) -> str:
    """Downloads the Wikipedia article data for the given title, stores it on disk and returns it"""
    try:
        article_data: str = retrieve_wikipedia_article(article_title)
        safe_name: str = article_title.replace("/", "_")
        # If the file already exists, return it from the file
        if os.path.exists(f"{article_folder}/{safe_name}.txt"):
            with open(f"{article_folder}/{safe_name}.txt", "r", encoding="utf-8") as file:
                return file.read()

        # Otherwise, download and store it
        with open(f"{article_folder}/{safe_name}.txt", "w", encoding="utf-8") as file:
            file.write(article_data)
        return article_data
    except Exception as e:
        print(f"Error while storing article data: {e}")
        raise
```

```python
def load_countries_by_continent(continent: str = "Europe", only_sovereign: bool = True) -> gpd.GeoDataFrame:
    """Returns GeoDataFrame containing countries from a specified continent with their names and polygons"""
    url: str = "https://naciscdn.org/naturalearth/10m/cultural/ne_10m_admin_0_countries.zip"
    response: Response = requests.get(url)

    with tempfile.TemporaryDirectory() as tmpdir:
        with ZipFile(BytesIO(response.content)) as zip_file:
            zip_file.extractall(tmpdir)

        shp_file = [f for f in os.listdir(tmpdir) if f.endswith('.shp')][0]
        shp_path = os.path.join(tmpdir, shp_file)

        gdf: gpd.GeoDataFrame = gpd.read_file(shp_path)

    gdf.columns = gdf.columns.str.lower()

    # Filter by continent
    filtered_gdf: gpd.GeoDataFrame = cast(gpd.GeoDataFrame, gdf[gdf['continent'] == continent].copy())

    if only_sovereign:
        # Keep only sovereign countries
        filtered_gdf = cast(gpd.GeoDataFrame, filtered_gdf[filtered_gdf['note_adm0'].isnull()])

    filtered_gdf = cast(gpd.GeoDataFrame, filtered_gdf[['name', 'continent', 'geometry']])

    filtered_gdf.reset_index(drop=True, inplace=True)

    return filtered_gdf
```

```python
def load_us_states(include_territories: bool = False) -> gpd.GeoDataFrame:
    """Returns GeoDataFrame containing US states with their names and polygons"""
    url: str = "https://naciscdn.org/naturalearth/10m/cultural/ne_10m_admin_1_states_provinces.zip"
    response: Response = requests.get(url)

    with tempfile.TemporaryDirectory() as tmpdir:
        with ZipFile(BytesIO(response.content)) as zip_file:
            zip_file.extractall(tmpdir)

        shp_file: str = [f for f in os.listdir(tmpdir) if f.endswith('.shp')][0]
        shp_path: str = os.path.join(tmpdir, shp_file)

        gdf: gpd.GeoDataFrame = gpd.read_file(shp_path)

    filtered_gdf: gpd.GeoDataFrame = cast(gpd.GeoDataFrame, gdf[gdf['iso_a2'] == 'US'].copy())

    if not include_territories:
        territories: list[str] = ['Puerto Rico', 'Guam', 'US Virgin Islands', 'American Samoa', 'Northern Mariana Islands']
        filtered_gdf = cast(gpd.GeoDataFrame, filtered_gdf[~filtered_gdf['name'].isin(territories)])

    filtered_gdf = cast(gpd.GeoDataFrame, filtered_gdf[['name', 'geometry']])

    filtered_gdf.reset_index(drop=True, inplace=True)

    return filtered_gdf
```

```python
def normalize_territory_name(territory_name: str) -> str:
    """Applies normalization on a territory name so that it is a valid Wikipedia article name"""
    spaces_cleaned: str = territory_name.strip().replace(" ", "_")
    if spaces_cleaned in TERRITORY_ARTICLE_MAPPING:
        return TERRITORY_ARTICLE_MAPPING[spaces_cleaned]
    return spaces_cleaned

def download_articles_for_dataframe(dataframe: gpd.GeoDataFrame) -> gpd.GeoDataFrame:
    """Downloads the Wikipedia article for each teritorry entry in the received dataframe and returns the updated dataframe containing that column"""
    def fetch_article(country_name: str) -> str | None:
        try:
            mapped_name: str = normalize_territory_name(country_name)
            article_data: str = store_article_data(mapped_name)
            return article_data
        except Exception as e:
            print(f"Error while downloading data for {country_name}: {e}")
            return None

    tqdm.pandas(desc="Downloading articles")
    dataframe['article'] = dataframe['name'].progress_apply(fetch_article)

    return dataframe
```

## Creating embeddings for the data

Code for creating embeddings for each data teritorry based on the Wikipedia article data

```python
def load_embedding_model(model_name_primary: str = MODEL_NAME_PRIMARY, fallback: str = MODEL_NAME_FALLBACK):
    used: str = model_name_primary
    try:
        logger.info("Loading embedding model: %s", model_name_primary)
        model = SentenceTransformer(model_name_primary)
    except Exception as e:
        logger.warning("Failed to load %s, falling back to %s", model_name_primary, fallback, e)
        model = SentenceTransformer(fallback)
        used = fallback
    return model, used
```

```python
def compute_document_embedding(text: str, model: SentenceTransformer) -> np.ndarray:
    max_tokens: int = min(MAX_TOKENS, model.max_seq_length)

    tokens: list[int] = model.tokenizer.encode(text, add_special_tokens=False, truncation=False)  # type: ignore[attr-defined]

    chunks: list[str] = []
    for index in range(0, len(tokens), max_tokens):
        chunk_tokens: list[int] = tokens[index:index + max_tokens]
        chunk_text: str = model.tokenizer.decode(chunk_tokens, skip_special_tokens=True)  # type: ignore[attr-defined]
        chunks.append(chunk_text)

    if not chunks:
        raise ValueError("Empty text for embedding")

    chunk_embs: np.ndarray = model.encode(chunks, convert_to_numpy=True, show_progress_bar=False, batch_size=4)
    avg: np.ndarray = np.mean(chunk_embs, axis=0)
    norm: float = float(np.linalg.norm(avg))
    if norm > 0:
        avg = avg / norm

    return avg
```

```python
def compute_all_embeddings(dataframe: gpd.GeoDataFrame, model: SentenceTransformer) -> gpd.GeoDataFrame:
    """Computes the embeddings for all entries in the received dataframe and returns the updated dataframe containing the new column"""
    def compute_embedding(article: str) -> np.ndarray | None:
        try:
            return compute_document_embedding(article, model)
        except Exception as e:
            logger.warning("Error while computing embedding", e)
            return None

    tqdm.pandas(desc="Computing embeddings")
    dataframe['embedding'] = dataframe['article'].progress_apply(compute_embedding)

    return dataframe
```

```python
def find_closest_matches(source: gpd.GeoDataFrame, target: gpd.GeoDataFrame) -> gpd.GeoDataFrame:
    """Finds the closest match in target for each entry in source based on embeddings"""
    target_names: list[str] = target['name'].tolist()
    target_embeddings: np.ndarray = np.vstack(target['embedding'].tolist())

    def find_match(source_embedding: np.ndarray) -> tuple[str, float]:
        source_emb_reshaped: np.ndarray = source_embedding.reshape(1, -1)
        similarities: np.ndarray = cosine_similarity(source_emb_reshaped, target_embeddings).reshape(-1)
        best_idx: int = int(np.argmax(similarities))
        return target_names[best_idx], float(similarities[best_idx])

        tqdm.pandas(desc="Finding closest matches")

    matches: list[tuple[str, float]] = source['embedding'].progress_apply(find_match).tolist()

    source['closest_match'] = [match[0] for match in matches]
    source['closest_similarity_score'] = [match[1] for match in matches]

    return source
```

```python
def main_data_gatherer() -> tuple[gpd.GeoDataFrame, gpd.GeoDataFrame]:
    # Download the data we need
    create_directories()
    europe = download_articles_for_dataframe(load_countries_by_continent('Europe'))
    us_states = download_articles_for_dataframe(load_us_states())

    # Compute the embeddings
    embedding_model, used_model = load_embedding_model()
    europe = compute_all_embeddings(europe, embedding_model)
    us_states = compute_all_embeddings(us_states, embedding_model)

    # Find the closest territories
    europe = find_closest_matches(europe, us_states)
    us_states = find_closest_matches(us_states, europe)

    # Save
    europe.to_file("europe_data.geojson", driver="GeoJSON")
    us_states.to_file("us_states_data.geojson", driver="GeoJSON")

    return europe, us_states

europe, us_states = main_data_gatherer()
```

## Creating the maps

Code for creating the simple maps that display the data. I didn't know the best way to do that so I have the code both for the text-overlay and the coloured variant

```python
# Visual styling
COUNTRY_COLOR = '#dda15e'  # Smooth Brown, autumn theme ¯\_(ツ)_/¯
BORDER_COLOR = 'white'
BORDER_WIDTH = 0.8
LABEL_FONT_SIZE = 12
LABEL_FONT_COLOR = '#000000'
LABEL_FONT_FAMILY = 'Arial, sans-serif'

# Figure dimensions
FIGURE_WIDTH = 1000
FIGURE_HEIGHT = 700
TITLE_TEXT = "European Countries by Nearest US State Wikipedia Description"
TITLE_FONT_SIZE = 24

# Projection settings
PROJECTION_TYPE = 'conic conformal'
PROJECTION_LON = 16.5
PROJECTION_LAT = 20
PROJECTION_SCALE = 1.6

# The default name / label position clashes with some neighbours, so we need to adjust them
# by specifying the exact coordinates for them
MANUAL_COORDS: dict[str, tuple[float, float]] = {
    'Austria': (47.5, 14.0),
    'United Kingdom': (52.7, -1.6),
    'Norway': (60.5, 8.5),
    'Lithuania': (55.75, 23.68),
    'Finland': (62.0, 26.0),
    'Russia': (55.75, 37.61),
    'Greece': (39.0, 22.0),
    'Italy': (42.5, 12.5),
    'Sweden': (60.0, 15.0),
    'France': (46.2, 2.2),
    'Ireland': (52.84, -7.78),
    'Belarus': (53.57, 27.76),
    'Estonia': (58.96, 25.20),
    'Denmark': (55.71, 9.03),
    'Croatia': (45.68, 16.05),
    'Serbia': (44.4, 20.9),
    # Add more countries here if their labels are misplaced
}

# We also need to separate micronations or smaller states, since they cannot be easily represented
MICRONATION_NAMES: list[str] = [
    'Monaco',
    'Malta',
    'Liechtenstein',
    'San Marino',
    'Andorra',
    'Luxembourg',
    'Vatican',
    'North Macedonia',
    'Kosovo',
    'Montenegro',
]

def prepare_europe_data(
    europe_df: gpd.GeoDataFrame,
    manual_coords: dict[str, tuple[float, float]] = MANUAL_COORDS
) -> gpd.GeoDataFrame:
    europe = europe_df.copy()

    # Now we compute the label position as the representative point for the country
    europe['label_point'] = europe.geometry.representative_point()
    europe['label_lon'] = europe['label_point'].x
    europe['label_lat'] = europe['label_point'].y

    # Also manual corrections are applied after this
    for country_name, (lat, lon) in manual_coords.items():
        country_rows = europe[europe['name'] == country_name]

        if not country_rows.empty:
            idx = country_rows.index[0]
            europe.loc[idx, 'label_lat'] = lat
            europe.loc[idx, 'label_lon'] = lon

    return europe

def create_micronation_legend(
    europe_df: gpd.GeoDataFrame,
    micronation_names: list[str] = MICRONATION_NAMES
) -> str:
    # Here is the legend text for the micronations
    micronations_europe = europe_df[europe_df['name'].isin(micronation_names)]
    legend_text = "<b>Other Matches:</b><br>"
    for _, row in micronations_europe.sort_values(by='name', ascending=True).iterrows():
        legend_text += f"{row['name']}: {row['closest_match']}<br>"

    return legend_text

def create_europe_map(
    europe_df: gpd.GeoDataFrame,
    micronation_names: list[str] = MICRONATION_NAMES
) -> go.Figure:
    # Now the drawing itself
    fig = go.Figure()

    # Layer with all country shapes
    fig.add_trace(go.Choropleth(
        geojson=europe_df.geometry.__geo_interface__,
        locations=europe_df.index,
        z=europe_df.index,
        colorscale=[[0, COUNTRY_COLOR], [1, COUNTRY_COLOR]],
        showscale=False,
        hoverinfo='none',
        marker=dict(
            line=dict(
                color=BORDER_COLOR,
                width=BORDER_WIDTH
            )
        )
    ))

    # Layer with the matching names
    europe_for_labels = europe_df[~europe_df['name'].isin(micronation_names)]
    fig.add_trace(go.Scattergeo(
        lat=europe_for_labels['label_lat'],
        lon=europe_for_labels['label_lon'],
        mode='text',
        text=europe_for_labels['closest_match'],
        textfont=dict(
            size=LABEL_FONT_SIZE,
            color=LABEL_FONT_COLOR,
            family=LABEL_FONT_FAMILY,
            weight='bold'
        ),
        hoverinfo='none'
    ))

    # Tooltip layer
    fig.add_trace(go.Scattergeo(
        lat=europe_df['label_lat'],
        lon=europe_df['label_lon'],
        mode='markers',
        marker=dict(size=5, opacity=0),
        text=europe_df.apply(lambda r: f"<b>{r['name']}</b><br>Match: {r['closest_match']}", axis=1),
        hoverinfo='text'
    ))

    # Create micronation legend
    legend_text = create_micronation_legend(europe_df, micronation_names)

    # Overall layout
    fig.update_layout(
        width=FIGURE_WIDTH,
        height=FIGURE_HEIGHT,
        title_text=TITLE_TEXT,
        title_font_size=TITLE_FONT_SIZE,
        title_x=0.5,
        showlegend=False,
        margin={"r":0, "t":50, "l":0, "b":0},

        # Set a conic projection, since it's one of the better ones for plain Europe
        geo=dict(
            scope='europe',
            projection=dict(
                type=PROJECTION_TYPE,
                rotation=dict(lon=PROJECTION_LON, lat=PROJECTION_LAT),
                scale=PROJECTION_SCALE
            ),
            visible=False,
            bgcolor='rgba(0,0,0,0)',
            showland=False,
            showcountries=False,
        ),

        # Add the micronation legend
        annotations=[
            dict(
                text=legend_text,
                align='left',
                showarrow=False,
                xref='paper', yref='paper',
                x=0.02, y=0.99,
                bgcolor='rgba(255, 255, 255, 0.85)',
                bordercolor='black',
                borderwidth=1,
                font=dict(size=11)
            )
        ]
    )

    return fig

europe_prepared: gpd.GeoDataFrame = prepare_europe_data(europe)
fig: go.Figure = create_europe_map(europe_prepared)
fig.show()
```

```python
# Visual styling
BORDER_COLOR = 'black'
BORDER_WIDTH = 0.8
STAR_SIZE = 15
STAR_BORDER_WIDTH = 1.5

# Figure dimensions
FIGURE_WIDTH = 1000
FIGURE_HEIGHT = 700
TITLE_TEXT = "European Countries by Nearest US State Wikipedia Description"
TITLE_FONT_SIZE = 24

# Projection settings
PROJECTION_TYPE = 'conic conformal'
PROJECTION_LON = 16.5
PROJECTION_LAT = 20
PROJECTION_SCALE = 1.6

# Micronations list
MICRONATION_NAMES: list[str] = [
    'Monaco', 'Malta', 'Liechtenstein', 'San Marino', 'Andorra', 'Luxembourg', 'Vatican'
]

# Manual coordinates for micronations (for star placement)
MICRONATION_COORDS: dict[str, tuple[float, float]] = {
    'Monaco': (43.73, 7.42),
    'Malta': (35.90, 14.51),
    'Liechtenstein': (47.14, 9.52),
    'San Marino': (43.94, 12.46),
    'Andorra': (42.51, 1.52),
    'Luxembourg': (49.61, 6.13),
    'Vatican': (41.90, 12.45),
}

def generate_colours(n: int) -> list[str]:
    # Use colours from Plotly to generate a random colour list
    colours = (
        px.colors.qualitative.Plotly +
        px.colors.qualitative.D3 +
        px.colors.qualitative.G10 +
        px.colors.qualitative.T10 +
        px.colors.qualitative.Alphabet
    )

    # Cycle through if we need more colors than available
    return [colours[i % len(colours)] for i in range(n)]

def prepare_europe_coloured_data(
    europe_df: gpd.GeoDataFrame,
    micronation_names: list[str] = MICRONATION_NAMES,
    micronation_coords: dict[str, tuple[float, float]] = MICRONATION_COORDS
) -> tuple[gpd.GeoDataFrame, gpd.GeoDataFrame, dict[str, str], list[list[float | str]]]:
    europe = europe_df.copy()

    # Create the colour mapping by match
    unique_states = sorted(europe['closest_match'].unique())
    n_states = len(unique_states)
    state_colours = generate_colours(n_states)
    colour_map = dict(zip(unique_states, state_colours))

    # Create numeric mapping for choropleth
    state_to_num = {state: i for i, state in enumerate(unique_states)}
    europe['state_num'] = europe['closest_match'].map(state_to_num)

    # Create the colour scale
    colourscale = []
    for i, state in enumerate(unique_states):
        colourscale.append([i / (n_states - 1) if n_states > 1 else 0, colour_map[state]])
        if i < n_states - 1:
            colourscale.append([(i + 0.99) / (n_states - 1), colour_map[state]])

    # Separate the micronations
    europe_micro = europe[europe['name'].isin(micronation_names)].copy()
    europe_micro['star_lat'] = europe_micro['name'].map(lambda x: micronation_coords.get(x, (0, 0))[0])
    europe_micro['star_lon'] = europe_micro['name'].map(lambda x: micronation_coords.get(x, (0, 0))[1])

    return europe, europe_micro, colour_map, colourscale


def create_colour_legend(colour_map: dict[str, str]) -> str:
    # Create the match legend
    legend_html = "<b>US State Matches:</b><br>"
    for state in sorted(colour_map.keys()):
        color = colour_map[state]
        legend_html += f'<span style="color:{color}">⬤</span> {state}<br>'

    return legend_html

def create_europe_colored_map(
    europe_df: gpd.GeoDataFrame,
    micronation_names: list[str] = MICRONATION_NAMES,
    micronation_coords: dict[str, tuple[float, float]] = MICRONATION_COORDS
) -> go.Figure:
    # Prepare data
    europe, europe_micro, colour_map, colorscale = prepare_europe_coloured_data(
        europe_df, micronation_names, micronation_coords
    )

    fig = go.Figure()

    # Layer with the colour for all countries
    fig.add_trace(go.Choropleth(
        geojson=europe.geometry.__geo_interface__,
        locations=europe.index,
        z=europe['state_num'],
        colorscale=colorscale,
        showscale=False,
        marker=dict(line=dict(color=BORDER_COLOR, width=BORDER_WIDTH)),
        hovertemplate='<b>%{customdata[0]}</b><br>Match: %{customdata[1]}<extra></extra>',
        customdata=europe[['name', 'closest_match']].values
    ))

    # Add stars for micronations
    fig.add_trace(go.Scattergeo(
        lat=europe_micro['star_lat'],
        lon=europe_micro['star_lon'],
        mode='markers',
        marker=dict(
            size=STAR_SIZE,
            color=[colour_map[match] for match in europe_micro['closest_match']],
            symbol='star',
            line=dict(color=BORDER_COLOR, width=STAR_BORDER_WIDTH)
        ),
        text=europe_micro.apply(lambda r: f"<b>{r['name']}</b><br>Match: {r['closest_match']}", axis=1),
        hoverinfo='text',
        showlegend=False
    ))

    # Overall layout
    legend_html = create_colour_legend(colour_map)
    fig.update_layout(
        width=FIGURE_WIDTH,
        height=FIGURE_HEIGHT,
        title_text=TITLE_TEXT,
        title_font_size=TITLE_FONT_SIZE,
        title_x=0.5,
        showlegend=False,
        margin={"r":0, "t":50, "l":0, "b":0},

        geo=dict(
            scope='europe',
            projection=dict(
                type=PROJECTION_TYPE,
                rotation=dict(lon=PROJECTION_LON, lat=PROJECTION_LAT),
                scale=PROJECTION_SCALE
            ),
            visible=False,
            bgcolor='rgba(0,0,0,0)',
            showland=False,
            showcountries=False,
        ),

        # Add the legend here
        annotations=[
            dict(
                text=legend_html,
                align='left',
                showarrow=False,
                xref='paper', yref='paper',
                x=0.02, y=0.99,
                bgcolor='rgba(255, 255, 255, 0.9)',
                bordercolor='black',
                borderwidth=1,
                font=dict(size=10),
                xanchor='left',
                yanchor='top'
            )
        ]
    )

    return fig

fig: go.Figure = create_europe_colored_map(europe)
fig.show()
```
